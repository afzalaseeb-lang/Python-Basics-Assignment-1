```python
s1="Hello"
s2=input("Enter Your Name: ")
s3=", Welcome To Python Programming"
s4= s1+" "+s2+s3
print(s4)
# Print the first character
print("First character:", s4[0])

# Print the last character
print("Last character:", s4[-1])

# Print the first 5 characters
print("First 5 characters:", s4[:5])

# Print the last 11 characters
print("Last 11 characters:", s4[-11:])

# Print the string in reverse
print("Reversed string:", s4[::-1])

# Print the word Python using slicing
start = s4.index("Python")
print(s4[start:start+6])
```

```python
# String Methods

strM = "Python beginner tutorial"

# Convert the sentence to uppercase
print("Uppercase:", strM.upper())

# Convert the sentence to lowercase
print("Lowercase:", strM.lower())

# Capitalize the sentence
print("Capitalized:", strM.capitalize())

# Count occurrences of character 't'
print("Occurrences of 't':", strM.count('t'))

# Replace "Python" with "Machine Learning"
print("Replaced string:", strM.replace("Python", "Machine Learning"))
```

```python

# Tuples: Creation, Concatenation and Access

# Create two tuples
t1 = (10, 20, 30)
t2 = (40, 50, 60)

# Concatenate the two tuples
t_combine = t1 + t2
print("Combined tuple:", t_combine)

# Repeat the elements 3 times
print("Repeated tuple:", t_combine * 3)

# Access the 3rd element
print("Third element:", t_combine[2])

# Access the first three elements
print("First three elements:", t_combine[:3])

# Access the last three elements
print("Last three elements:", t_combine[-3:])
```
